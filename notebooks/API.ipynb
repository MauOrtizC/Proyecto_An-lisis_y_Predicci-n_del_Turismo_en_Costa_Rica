
## Importar dependencias


```python

import sys, os
project_root = os.path.abspath("..")  # raíz del proyecto
sys.path.append(project_root)

from src.api.API import ClienteAPI


```


## Crear instancia y obtener datos diarios


```python

api = ClienteAPI()

# Obtener datos diarios (temperatura máxima, mínima y lluvia)
df_diario = api.obtener_datos_climaticos_diarios(
    lat=9.9, lon=-84.2,
    start_date="2024-01-01",
    end_date="2024-12-31",
    variables=["temperature_2m_max", "temperature_2m_min", "precipitation_sum"]
)

# Guardar datos diarios en data/raw/
api.guardar_csv(df_diario, "clima_costa_rica_2024_diario")
```


## Convertir a valores MENSUALES


```python

df_mensual = api.convertir_a_mensual(df_diario)

# Guardar datos mensuales en data/raw/
api.guardar_csv(df_mensual, "clima_costa_rica_2024_mensual")

# =====================
# BLOQUE 4: Vista previa
# =====================
print(df_mensual.head())

```
